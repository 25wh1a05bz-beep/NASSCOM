```python
import pandas as pd

df = pd.DataFrame({
    'name': ['A', 'B', 'C'],
    'age': [20, 25, None],
    'score': [85, 90, 78]
})

df.info()
```

```python


print('Missing per column:')
print(df.isna().sum())

print('\nDuplicate rows:', df.duplicated().sum())

print('\nColumns:')
print(df.columns.tolist())

if 'spend' in df.columns:
    print('\nNote: spend is type', df['spend'].dtype, '-> stored as text!')
else:
    print("\n'spend' column not found.")
```
