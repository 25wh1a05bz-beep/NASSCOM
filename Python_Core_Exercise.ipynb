```python
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}
```

```python
import random
from datetime import datetime, timedelta

# Example values (define these if not already provided)
DEVICE_TYPES = ["Temperature", "Humidity", "Pressure", "Light"]

LOCATIONS = [
    "Lab A",
    "Lab B",
    "Warehouse",
    "Office",
    "Factory"
]

READING_RANGES = {
    "Temperature": (15.0, 40.0),
    "Humidity": (20.0, 90.0),
    "Pressure": (950.0, 1050.0),
    "Light": (100.0, 1000.0)
}

raw_sensor_data = []

# Starting timestamp
start_time = datetime.now()

for i in range(100):
    timestamp = (start_time + timedelta(seconds=i)).strftime("%Y-%m-%d %H:%M:%S")
    device_id = f"DEV{i+1:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    reading = (timestamp, device_id, device_type, location, value)
    raw_sensor_data.append(reading)

# Display first 5 readings
for reading in raw_sensor_data[:5]:
    print(reading)
```

```python
temperature_readings = []

for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading

    if device_type == "Temperature_Sensor":
        temperature_readings.append({
            "timestamp": timestamp,
            "device_id": device_id,
            "location": location,
            "temperature_celsius": value
        })

# Display first 5 temperature readings
for temp in temperature_readings[:5]:
    print(temp)
```

```python
unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print(unique_locations)
```

```python
device_stats = {}

for reading in raw_sensor_data:
    _, _, device_type, _, value = reading

    if device_type not in device_stats:
        # Initialize with the first value
        device_stats[device_type] = (value, value)
    else:
        min_value, max_value = device_stats[device_type]
        device_stats[device_type] = (
            min(min_value, value),
            max(max_value, value)
        )

print(device_stats)
```

```python
def get_readings_by_location(readings_list, target_location):
    filtered_readings = []

    for reading in readings_list:
        location = reading[3]  # Location is the 4th element in the tuple

        if location == target_location:
            filtered_readings.append(reading)

    return filtered_readings


# Example usage:
location_readings = get_readings_by_location(raw_sensor_data, "Lab A")

print(location_readings[:5])
```

```python
def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total_temperature = 0.0

    for reading in temp_data_list:
        total_temperature += reading["temperature_celsius"]

    average_temperature = total_temperature / len(temp_data_list)

    return average_temperature



average_temperature = calculate_average_temperature(temperature_readings)

print("Average Temperature:", average_temperature)
```

```python
def generate_report_line(reading_dict):
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Example usage:
if temperature_readings:
    report_line = generate_report_line(temperature_readings[0])
    print(report_line)
```

```python
import os

# Create directory if it does not exist
reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"\nDirectory '{reports_dir}' created.")
else:
    print(f"\nDirectory '{reports_dir}' already exists.")
```

```python
import datetime

if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]


    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)


    duration = last_dt - first_dt

    days = duration.days
    total_seconds = duration.total_seconds()

    hours = int((total_seconds % 86400) // 3600)
    minutes = int((total_seconds % 3600) // 60)

    print(f"\nDuration of data collection: {days} days, {hours} hours, {minutes} minutes.")

else:
    print("\nNot enough data to calculate duration.")
```
